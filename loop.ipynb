```python
# loop 4
peserta = ["ijul", "chlea", "reza","citra", "budi"]
for i in peserta:
  if i == "reza":
    continue
  print(i)
```

```python
# loop 5
departemen = ["statbis", "statistika bisnis", "matematika"]
angkatan = [2025, 2026]
for a in departemen:
  for b in angkatan:
    print(f"{a} angkatan {b}")
```

```python
# loop 6
pelamar = [("Budi", 3.2, 5),("Siti", 3.6, 3), ("Dodi", 3.8, 1)]

print("=== Hasil Pelamar Asisten ===")
for nama, IPK, semester in pelamar:
  if IPK >= 3.5 and semester > 2:
    status = "DITERIMA!!"
  else:
    status = "DITOLAK!"
  print(f"Nama: {nama} | Status: {status}")
```

```python
# loop 7
karyawan = [("Joko", "IT", 7000), ("Rina", "HR", 5000),
            ("Toni", "Data", 8000), ("Toni", "IT", 6500)]

total_gaji = 0
print("=== Total Gaji Karyawan ===")
for nama,departemen,gaji in karyawan:
  if departemen == "IT" or departemen == "Data":
    total_gaji += gaji

print("-" * 30)
print(f"Total Gaji: Rp{total_gaji}")
```

```python
# loop 8
penjualan = [("Kopi", 60, False), ("Teh", 30, False), ("Boba", 80, True)]
for nama_minuman, jumlah_penjualan, sedang_promo in penjualan:
  if jumlah_penjualan > 59 and not sedang_promo:
    print(nama_minuman)
```
